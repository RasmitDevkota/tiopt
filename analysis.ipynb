```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
file = open("V.txt", "r")
arr = np.array([[float(val) for val in line.strip("\n\r").split(",")] for line in file.readlines()])
print(arr)
print(arr.shape)
```

```python
xs = arr[:, 0]
ys = arr[:, 1]
zs = arr[:, 2]
Vs = arr[:, 3]
```

```python
fig = plt.figure(dpi=150)
ax = fig.add_subplot(projection='3d')

skip = 1

cs = np.copy(Vs)

# mask = cs > -1E9
# mask = cs > 1E-3
mask = cs > 5E-1
cs = cs[mask]
mask2 = cs <= 2.0
xs = arr[:,0][mask][mask2][::skip]
ys = arr[:,1][mask][mask2][::skip]
zs = arr[:,2][mask][mask2][::skip]
cs = arr[:,3][mask][mask2][::skip]
ss = (cs-min(cs))/(max(cs)-min(cs))

scatter = ax.scatter(xs, ys, zs, c=cs, s=ss, norm="linear")
plt.colorbar(scatter)

plt.title("Electric potential (V) vs. position")

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
```
